```python
import rclpy as rp
from geometry_msgs.msg import Twist
from turtlesim.msg import Pose

if not rp.ok():
    rp.init()

test_node = rp.create_node('pub_test')
pub = test_node.create_publisher(Twist, '/turtle1/cmd_vel', 10)

# Initial pose setup (orientation only)
pose = Pose()
pose.theta = 0.0

# Timer callback to publish continuously

def callback():
    msg = Twist()
    msg.linear.x = 2.5
    msg.angular.z = 2.5
    pub.publish(msg)


timer = test_node.create_timer(0.5, callback)
```
