```python
import rclpy
from turtlesim.srv import TeleportAbsolute

# Initialize ROS2
rclpy.init()
test_node = rclpy.create_node('service_client_node')
```

```python
# Create service client
service_name = '/turtle1/teleport_absolute'
cli = test_node.create_client(TeleportAbsolute, service_name)
```

```python
# Wait for service to be available
while not cli.wait_for_service(timeout_sec=1.0):
    print(f'Service {service_name} not available, waiting again...')

# Create request
req = TeleportAbsolute.Request()
req.x = 3.0
req.y = 3.0
req.theta = 0.0

# Call service asynchronously
future = cli.call_async(req)

# Spin once to process the response
rclpy.spin_once(test_node)

# Get the response
response = future.result()
print(f'Response: {response}')
```

```python
# Continue with another service call
req.x = float(2)
req.y = 3.0
future = cli.call_async(req)

while not future.done():
    rclpy.spin_once(test_node)

print(future.done(), future.result())
```
